```python
from pathlib import Path

BASE = str(Path.cwd() / "VOXINSIGHT_v1_output")

dirs = [
    f"{BASE}/features",
    f"{BASE}/features/mfcc",
    f"{BASE}/features/melspec",
    f"{BASE}/models",
    f"{BASE}/results",
    f"{BASE}/results/plots",
    f"{BASE}/results/checkpoints",
]

for d in dirs:
    Path(d).mkdir(parents=True, exist_ok=True)
    print(f"Created: {d}")

print("\nFolder structure ready.")
```

```python
from pathlib import Path

NOTEBOOK_ROOT = Path.cwd()
RAVDESS_ROOT = NOTEBOOK_ROOT / "audio_speech_actors_01-24"
CREMAD_ROOT  = NOTEBOOK_ROOT / "AudioWAV"

print("=== RAVDESS ===")
if RAVDESS_ROOT.exists():
    for actor_folder in sorted(RAVDESS_ROOT.iterdir())[:3]:
        if actor_folder.is_dir():
            files = list(actor_folder.iterdir())
            print(f"  {actor_folder.name}/  ({len(files)} files)")
else:
    print(f"  Missing folder: {RAVDESS_ROOT}")

print("\n=== CREMA-D ===")
if CREMAD_ROOT.exists():
    cremad_files = [f for f in CREMAD_ROOT.rglob("*.wav")]
    print(f"  Total .wav files found: {len(cremad_files)}")
    if cremad_files:
        print(f"  Example: {cremad_files[0].name}")
else:
    print(f"  Missing folder: {CREMAD_ROOT}")
```
